***1 min***

```python
from twelvedata import TDClient
import pandas as pd

td = TDClient(apikey="564594893d244679bc75c292bb2f4949")

symb = "AAPL"
interval = "1min"
first_start = "2026-01-08 16:50:00"
first_end = "2026-01-28 15:39:00"
second_start = "2026-01-28 15:40:00"
second_end = "2026-02-13 20:59:00"
size = 5000
```

```python
ts1 = td.time_series(
    symbol=symb,
    interval=interval,
    outputsize=size,
    timezone="utc",
    start_date=first_start,
    end_date=first_end
)

ts1_data = ts1.as_pandas()
```

```python
ts2 = td.time_series(
    symbol=symb,
    interval=interval,
    outputsize=size,
    timezone="utc",
    start_date=second_start,
    end_date=second_end
)

ts2_data = ts2.as_pandas()
```

```python
res = pd.concat([ts2_data, ts1_data]).iloc[::-1]
res.to_csv("datasets/AAPL_1min.csv")
```
