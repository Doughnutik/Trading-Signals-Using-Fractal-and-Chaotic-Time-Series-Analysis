```python
import mplfinance as mpf
import pandas as pd

df_name = "AAPL_15min"
df = pd.read_csv(f'datasets/{df_name}_marked.csv')
df["datetime"] = pd.to_datetime(df["datetime"])
df = df.set_index("datetime")
df.info()
```

```python
df_copy = df[df.index.year > 2024]
```

```python
mpf.plot(
    df_copy[1500:2000],
    type='candle',
    style='charles',
    title=f'AAPL 1 hour prices',
    ylabel='Price ($)',
    volume=True,
    figsize=(18, 8)
)
```
