```python
import mplfinance as mpf
import pandas as pd

df = pd.read_csv('datasets/AAPL_5min.csv')
df["datetime"] = pd.to_datetime(df["datetime"])
df = df.set_index("datetime")
df.info()
```

```python
df.head(5)
```

```python
mpf.plot(
    df.head(100),
    type='candle',
    style='charles',
    title=f'AAPL 5 minute prices',
    ylabel='Price ($)',
    volume=True,
    figsize=(18, 8)
)
```
