```python
import pandas as pd

alpha = 0.001

def mark_min_points(df):
    df['is_min'] = 0
    for i in range(4, len(df) - 5):
        if df.iloc[i]['open'] > df.iloc[i]['close']:
            continue
        minL = min(min(df.iloc[i-4:i]['close']), min(df.iloc[i-4:i]['open']))
        minR = min(min(df.iloc[i+1:i+5]['open']), min(df.iloc[i+1:i+5]['close']))
        if df.iloc[i]['open'] > minL or df.iloc[i]['open'] > minR:
            continue
        if df.iloc[i]['open'] <= df.iloc[i-4]['open'] * (1 - alpha) and df.iloc[i]['open'] <= df.iloc[i+4]['close'] * (1 - alpha):
            df.loc[df.index[i], "is_min"] = 1
    return df

def mark_max_points(df):
    df['is_max'] = 0
    for i in range(4, len(df) - 5):
        if df.iloc[i]['open'] < df.iloc[i]['close']:
            continue
        maxL = max(max(df.iloc[i-4:i]['close']), max(df.iloc[i-4:i]['open']))
        maxR = max(max(df.iloc[i+1:i+5]['open']), max(df.iloc[i+1:i+5]['close']))
        if df.iloc[i]['open'] < maxL or df.iloc[i]['open'] < maxR:
            continue
        if df.iloc[i]['open'] >= df.iloc[i-4]['open'] * (1 + alpha) and df.iloc[i]['open'] >= df.iloc[i+4]['close'] * (1 + alpha):
            df.loc[df.index[i], "is_max"] = 1
    return df
```

```python
df_name = "AAPL_1min"
df = pd.read_csv(f'../datasets/{df_name}.csv')
df["datetime"] = pd.to_datetime(df["datetime"])
df = df.set_index("datetime")
df.shape
```

```python
df = mark_min_points(df)
df = mark_max_points(df)
df[(df['is_min'] == 1) | (df['is_max'] == 1)]
```

```python
df.to_csv(f'../datasets/{df_name}_marked.csv')
```
