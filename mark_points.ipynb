```python
import pandas as pd

alpha = 0.005

def mark_points(df):
    df['is_min'] = 0
    for i in range(4, len(df) - 5):
        if df.iloc[i]['open'] > df.iloc[i]['close']:
            continue
        minL = min(min(df.iloc[i-4:i]['close']), min(df.iloc[i-4:i]['open']))
        minR = min(min(df.iloc[i+1:i+5]['open']), min(df.iloc[i+1:i+5]['close']))
        if df.iloc[i]['open'] > minL or df.iloc[i]['open'] > minR:
            continue
        if df.iloc[i]['open'] <= df.iloc[i-4]['open'] * (1 - alpha) and df.iloc[i]['open'] <= df.iloc[i+4]['close'] * (1 - alpha):
            df.loc[df.index[i], "is_min"] = 1
    return df
```

```python
df = pd.read_csv('datasets/AAPL_15min.csv')
df["datetime"] = pd.to_datetime(df["datetime"])
df = df.set_index("datetime")
df = mark_points(df)

df = df[df['is_min'] == 1]

df
```
